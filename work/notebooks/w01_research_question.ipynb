## My Lane: Content Refresh Prioritization

I chose the Content Refresh lane because every website faces the same
problem: which pages should we update first? Manual review of thousands
of pages is impossible. A data-driven approach can rank pages by
predicted decline risk, helping editors focus on what matters most.

This lane connects directly to real business value — refreshing the
right pages means better search rankings, more traffic, and higher
engagement. It's a classic supervised learning problem with clear
success metrics (Precision@50).

## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

## The Question: Which pages need content refresh first?

**Decision:** Content teams need to decide which pages to refresh
each week from thousands of candidates.

**Action:** A ranked queue is generated. Human editors review the
top-ranked pages and assign writers to refresh them. The model
suggests, humans decide.

**Cost of a Wrong Call:**
- **Missing a declining page (False Negative):** Traffic continues
  to drop. Competitors outrank us. Lost revenue.
- **Flagging a good page (False Positive):** Writer time wasted on
  a page that didn't need it. Opportunity cost — another page
  could have been refreshed instead.

**Why ML helps:** Hand-written rules achieved only 24% Precision@50.
ML can learn from multiple signals (impressions, clicks, position)
and achieve 90% Precision@50 — making the team 3.75x more efficient.

```python
import pandas as pd
import numpy as np

# Load starter data
df = pd.read_csv('data/raw/content_refresh_anonymized.csv')

print(f"✅ Data loaded! Shape: {df.shape}")
print(f"Columns: {df.columns.tolist()}")

# Number 1: How many pages are declining?
declining = df['trend_direction'].str.lower().eq('down').sum()
total = len(df)
print(f"\n1. Declining pages: {declining}/{total} ({declining/total*100:.1f}%)")
print(f"   → Nearly {declining/total*100:.0f}% of pages are declining — a real problem!")

# Number 2: Correlation between CTR and position
corr = df['ctr'].corr(df['avg_position'])
print(f"\n2. CTR vs Position correlation: {corr:.3f}")
print(f"   → Negative correlation confirms: better position = higher CTR")

# Number 3: Average content age
avg_age = df['content_age_days'].mean()
print(f"\n3. Average content age: {avg_age:.0f} days ({avg_age/365:.1f} years)")
print(f"   → Content is old — refresh opportunities exist!")
```
