## My Lane: Content Refresh Prioritization

### ML Task Type: **Ranking (Scoring)**

I'm not just classifying pages as "refresh" or "don't refresh" — I'm ranking them
by priority. The model should score each page from 0 (low priority) to 1 (urgent
refresh needed), so the content team can work through the queue in order.

This is a **supervised learning** problem: I have historical data showing which
pages declined, and I can learn patterns from that.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

## Target: Predicting Likelihood of Decline

My target is **trend_direction** — specifically identifying pages that are likely
to be declining. I'll create a binary label:

- `1` (Positive class): Page is declining — needs refresh
- `0` (Negative class): Page is stable or rising — no action needed

This is a **proxy** for "needs refresh" because:
- A declining trend is a measurable signal that the page is losing value
- Not all declining pages need refresh, but it's a strong indicator
- The content team can validate before taking action

## 3. Success metric

*One metric you can defend. What number means 'good'?*

## Success Metric: Precision@50

I chose **Precision@50** as my primary metric because:

- The content team can realistically review ~50 pages per week
- Precision measures "of the pages we flagged, how many actually needed refresh?"
- High precision = less wasted effort by the content team

Secondary metrics: Recall@50 (are we missing important pages?)

Current baseline (hand rule): ~24% Precision@50
Target (ML model): 70%+ Precision@50

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

## Why ML Beats a Hand-Written Rule

### Current Rule (Hand-Written):
If days_since_update > 180 AND impressions > 500 → Flag as "needs refresh"
Precision@50: ~24%

### Why ML is Better:

1. **Multiple signals combined:** ML weighs impressions, CTR, position, age, and word count simultaneously
2. **Learns thresholds automatically:** Instead of guessing "180 days" and "500 impressions," ML finds optimal values
3. **Adapts to patterns:** Different content types may have different "needs refresh" signals
4. **Scalable:** As more data comes in, the model can be retrained; rules need manual updates

### Evidence (from Week 1):
- Hand rule Precision@50: ~0.24
- Simple decision tree Precision@50: ~0.60 (2.5x improvement!)
